```python
dup_fac_ids = deduped[deduped.duplicated(subset=['facilityid'], keep=False)]['facilityid'].unique()

dups = deduped[deduped['facilityid'].isin(dup_fac_ids)].sort_values('facilityid')

from collections import defaultdict

uniqueness_causes = defaultdict(list)

columns_to_check = [col for col in deduped.columns if col != 'facilityid']

for facid, group in dups.groupby('facilityid'):
    diffs = [col for col in columns_to_check if group[col].nunique(dropna=False) > 1]
    if diffs:
        uniqueness_causes[facid] = diffs

for i, (facid, cols) in enumerate(uniqueness_causes.items()):
    print(f"facilityid {facid} differs by columns: {cols}")
    if i >= 4:
        break

print(f"\nNumber of facilityids with multiple rows due to differences in these columns: {len(uniqueness_causes)}")

```

```python
all_causing_columns = set()
for cols in uniqueness_causes.values():
    all_causing_columns.update(cols)
print("Set of all columns causing duplicity across affected facilityids:")
print(all_causing_columns)
```

```python
allindrawn_uniques = deduped.groupby('facilityid')['allindrawn'].nunique(dropna=False)

facilityids_multiple_allindrawn = allindrawn_uniques[allindrawn_uniques > 1]

if not facilityids_multiple_allindrawn.empty:
    print("Facilityids with more than one unique allindrawn value:")
    print(facilityids_multiple_allindrawn)
    for facid in facilityids_multiple_allindrawn.index[:5]:
        values = deduped.loc[deduped['facilityid'] == facid, 'allindrawn'].unique()
        print(f"facilityid {facid}: allindrawn values = {values}")
else:
    print("All facilityids have the same allindrawn value for all their rows.")

print(f"\nNumber of facilityids with differing allindrawn values: {len(facilityids_multiple_allindrawn)}")

```

```python
import os
import pandas as pd

macros_folder = 'macros'
macro_csv_files = [f for f in os.listdir(macros_folder) if f.endswith('.csv')]

macro_dfs = {}

for csv_file in macro_csv_files:
    file_path = os.path.join(macros_folder, csv_file)
    df_name = os.path.splitext(csv_file)[0]
    macro_dfs[df_name] = pd.read_csv(file_path)
```

```python
from functools import reduce

for k, df in macro_dfs.items():
    if 'observation_date' not in df.columns:
        raise ValueError(f"{k} does not have 'observation_date' column!")

macro_merged_df = reduce(
    lambda left, right: pd.merge(left, right, on='observation_date', how='outer'),
    macro_dfs.values()
)

macro_merged_df['observation_date'] = pd.to_datetime(macro_merged_df['observation_date'])
macro_merged_df = macro_merged_df[macro_merged_df['observation_date'].dt.year >= 1999]

macro_merged_df['year'] = macro_merged_df['observation_date'].dt.year

macro_yearly_df = macro_merged_df.groupby('year').mean(numeric_only=True).reset_index()

macro_yearly_df
```

```python
merged_final_df = pd.merge(merged_final_df, macro_yearly_df, on='year', how='left')
```

```python
from sklearn.linear_model import LassoCV
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
import numpy as np
import pandas as pd

X_train = traindf[features]
y_train = traindf['allindrawn']
X_test = testdf[features]
y_test = testdf['allindrawn']

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

lasso = LassoCV(cv=5, random_state=0, max_iter=5000)
lasso.fit(X_train_scaled, y_train)

y_train_pred = lasso.predict(X_train_scaled)
y_test_pred = lasso.predict(X_test_scaled)

print("Lasso Regression Results")
print(f"Train RMSE: {np.sqrt(mean_squared_error(y_train, y_train_pred)):.4f}")
print(f"Test RMSE: {np.sqrt(mean_squared_error(y_test, y_test_pred)):.4f}")
print(f"Train R^2: {r2_score(y_train, y_train_pred):.4f}")
print(f"Test R^2: {r2_score(y_test, y_test_pred):.4f}")

feat_imp = pd.Series(lasso.coef_, index=features)
print("Top 10 features by absolute coefficient:")
print(feat_imp.abs().sort_values(ascending=False).head(10))
```

```python
from xgboost import XGBRegressor
from sklearn.preprocessing import StandardScaler

safe_features = [f.replace('[','').replace(']','').replace('<','') for f in features]

scaler_xgb = StandardScaler()
X_train_safe = X_train.copy()
X_train_safe.columns = safe_features
X_test_safe = X_test.copy()
X_test_safe.columns = safe_features

X_train_safe_scaled = pd.DataFrame(
    scaler_xgb.fit_transform(X_train_safe), columns=safe_features, index=X_train_safe.index
)
X_test_safe_scaled = pd.DataFrame(
    scaler_xgb.transform(X_test_safe), columns=safe_features, index=X_test_safe.index
)

xgb = XGBRegressor(
    n_estimators=500,
    learning_rate=0.05,
    max_depth=4,
    subsample=0.7,
    colsample_bytree=0.7,
    random_state=42,
    n_jobs=-1,
    tree_method='hist',
)
xgb.fit(X_train_safe_scaled, y_train)

y_train_pred_xgb = xgb.predict(X_train_safe_scaled)
y_test_pred_xgb = xgb.predict(X_test_safe_scaled)

print("\nXGBoost Regression Results (with normalization)")
print(f"Train RMSE: {np.sqrt(mean_squared_error(y_train, y_train_pred_xgb)):.4f}")
print(f"Test RMSE: {np.sqrt(mean_squared_error(y_test, y_test_pred_xgb)):.4f}")
print(f"Train R^2: {r2_score(y_train, y_train_pred_xgb):.4f}")
print(f"Test R^2: {r2_score(y_test, y_test_pred_xgb):.4f}")

importances = pd.Series(xgb.feature_importances_, index=safe_features)
print("Top 10 features by importance:")
print(importances.sort_values(ascending=False).head(10))
```

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

# Set random seed for reproducibility
torch.manual_seed(42)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(42)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

scaler_X = StandardScaler()
X_train_scaled = scaler_X.fit_transform(X_train)
X_test_scaled = scaler_X.transform(X_test)

y_train_np = y_train.to_numpy() if hasattr(y_train, 'to_numpy') else y_train.values
y_test_np = y_test.to_numpy() if hasattr(y_test, 'to_numpy') else y_test.values
y_mean = np.mean(y_train_np)
y_std = np.std(y_train_np)
y_train_scaled = ((y_train_np - y_mean) / y_std).astype('float32')
y_test_scaled = ((y_test_np - y_mean) / y_std).astype('float32')

X_train_t = torch.tensor(X_train_scaled, dtype=torch.float32)
y_train_t = torch.tensor(y_train_scaled, dtype=torch.float32).view(-1, 1)
X_test_t = torch.tensor(X_test_scaled, dtype=torch.float32)
y_test_t = torch.tensor(y_test_scaled, dtype=torch.float32).view(-1, 1)

batch_size = 64
train_ds = TensorDataset(X_train_t, y_train_t)
test_ds = TensorDataset(X_test_t, y_test_t)
train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True)
test_loader = DataLoader(test_ds, batch_size=batch_size, shuffle=False)

class MLP(nn.Module):
    def __init__(self, input_dim, hidden_dims=[256, 128, 64], dropout=0.2):
        super().__init__()
        layers = []
        prev_dim = input_dim
        for h_dim in hidden_dims:
            layers.append(nn.Linear(prev_dim, h_dim))
            layers.append(nn.BatchNorm1d(h_dim))
            layers.append(nn.ReLU())
            layers.append(nn.Dropout(dropout))
            prev_dim = h_dim
        layers.append(nn.Linear(prev_dim, 1))
        self.model = nn.Sequential(*layers)
    def forward(self, x):
        return self.model(x)

mlp = MLP(X_train_t.shape[1], dropout=0.3).to(device)
criterion = nn.MSELoss()
optimizer = optim.AdamW(mlp.parameters(), lr=1e-3, weight_decay=1e-4)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode='min', factor=0.5, patience=10
)

epochs = 150
best_test_rmse = float('inf')
stopping_counter = 0
early_stop_patience = 25
best_weights = None

for epoch in range(1, epochs+1):
    mlp.train()
    train_loss = 0
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        output = mlp(xb)
        loss = criterion(output, yb)
        loss.backward()
        optimizer.step()
        train_loss += loss.item() * xb.size(0)
    train_loss /= len(train_loader.dataset)

    mlp.eval()
    with torch.no_grad():
        test_preds = []
        test_targets = []
        for xb, yb in test_loader:
            xb, yb = xb.to(device), yb.to(device)
            out = mlp(xb)
            test_preds.append(out.cpu())
            test_targets.append(yb.cpu())
        test_preds = torch.cat(test_preds, dim=0).numpy()
        test_targets = torch.cat(test_targets, dim=0).numpy()
        test_rmse = np.sqrt(np.mean((test_preds - test_targets) ** 2))

    scheduler.step(test_rmse)
    this_improved = test_rmse < best_test_rmse - 1e-5
    if this_improved:
        best_test_rmse = test_rmse
        stopping_counter = 0
        best_weights = mlp.state_dict()
    else:
        stopping_counter += 1

    if (epoch % 10 == 0) or (epoch == 1) or (epoch == epochs):
        print(f"Epoch {epoch}/{epochs} | Train MSE: {train_loss:.4f}, Test RMSE (scaled): {test_rmse:.4f}")
    if stopping_counter > early_stop_patience:
        print(f"Early stopping at epoch {epoch}, best test RMSE (scaled): {best_test_rmse:.4f}")
        break

if best_weights is not None:
    mlp.load_state_dict(best_weights)
else:
    print("Warning: best_weights was not set. Using last epoch weights (model may be suboptimal)!")

mlp.eval()
with torch.no_grad():
    y_train_pred_torch = mlp(X_train_t.to(device)).cpu().numpy().flatten() * y_std + y_mean
    y_test_pred_torch = mlp(X_test_t.to(device)).cpu().numpy().flatten() * y_std + y_mean

print("\nPyTorch MLP Regression Results (inverse scaled targets)")
print(f"Train RMSE: {np.sqrt(mean_squared_error(y_train_np, y_train_pred_torch)):.4f}")
print(f"Test RMSE: {np.sqrt(mean_squared_error(y_test_np, y_test_pred_torch)):.4f}")
print(f"Train R^2: {r2_score(y_train_np, y_train_pred_torch):.4f}")
print(f"Test R^2: {r2_score(y_test, y_test_pred_torch):.4f}")

```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error, r2_score

# Set up device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# Define activation functions mapping
activation_map = {
    'linear': nn.Identity,
    'sigmoid': nn.Sigmoid,
    'tanh': nn.Tanh,
    'relu': nn.ReLU
}

# Hyperparameter grid
hidden_sizes = [64, 128, 256]
learning_rates = [1e-3, 5e-4, 1e-4]
activations = ['linear', 'sigmoid', 'tanh', 'relu']

# Datasets: Use X_train_t, y_train_t, X_test_t, y_test_t as already prepared/scaled tensors
X_train_np = X_train_t.cpu().numpy()
y_train_np = y_train_t.cpu().numpy()
X_test_np = X_test_t.cpu().numpy()
y_test_np = y_test_t.cpu().numpy()

# Basic single-output MLP class
class SimpleMLP(nn.Module):
    def __init__(self, in_dim, hidden_dim, activation_name):
        super(SimpleMLP, self).__init__()
        self.activation = activation_map[activation_name]()
        self.hidden = nn.Linear(in_dim, hidden_dim)
        self.out = nn.Linear(hidden_dim, 1)

    def forward(self, x):
        x = self.activation(self.hidden(x))
        x = self.out(x)
        return x

results = []
training_loss_curves = {}

epochs = 100
early_stop_patience = 10  # Early stopping patience

input_dim = X_train_t.shape[1]

for hidden_dim in hidden_sizes:
    for lr in learning_rates:
        for act in activations:
            model = SimpleMLP(input_dim, hidden_dim, act).to(device)
            optimizer = optim.Adam(model.parameters(), lr=lr)
            criterion = nn.MSELoss()
            train_losses = []
            best_val_loss = float('inf')
            best_params = None
            stopping_counter = 0
            # Early stopping, 10% of train as "val"
            val_split = int(0.9 * X_train_np.shape[0])
            X_tn, X_val = X_train_np[:val_split], X_train_np[val_split:]
            y_tn, y_val = y_train_np[:val_split], y_train_np[val_split:]
            X_tn_t = torch.tensor(X_tn, dtype=torch.float32, device=device)
            y_tn_t = torch.tensor(y_tn, dtype=torch.float32, device=device)
            X_val_t = torch.tensor(X_val, dtype=torch.float32, device=device)
            y_val_t = torch.tensor(y_val, dtype=torch.float32, device=device)

            for epoch in range(1, epochs+1):
                # Train
                model.train()
                optimizer.zero_grad()
                pred_train = model(X_tn_t).flatten()
                loss = criterion(pred_train, y_tn_t.flatten())
                loss.backward()
                optimizer.step()
                train_losses.append(loss.item())
                
                # Validation
                model.eval()
                with torch.no_grad():
                    pred_val = model(X_val_t).flatten()
                    val_loss = criterion(pred_val, y_val_t.flatten()).item()
                if val_loss < best_val_loss - 1e-5:
                    best_val_loss = val_loss
                    best_params = model.state_dict()
                    stopping_counter = 0
                else:
                    stopping_counter += 1
                if stopping_counter > early_stop_patience:
                    break

            # Evaluate on test set with best weights
            if best_params is not None:
                model.load_state_dict(best_params)
            model.eval()
            with torch.no_grad():
                yhat_test = model(torch.tensor(X_test_np, dtype=torch.float32, device=device)).cpu().numpy().flatten()
                yhat_train = model(torch.tensor(X_train_np, dtype=torch.float32, device=device)).cpu().numpy().flatten()
            test_rmse = np.sqrt(mean_squared_error(y_test_np.flatten(), yhat_test))
            train_rmse = np.sqrt(mean_squared_error(y_train_np.flatten(), yhat_train))
            test_r2 = r2_score(y_test_np.flatten(), yhat_test)
            train_r2 = r2_score(y_train_np.flatten(), yhat_train)
            results.append({
                'hidden_dim': hidden_dim,
                'lr': lr,
                'activation': act,
                'final_train_loss': train_losses[-1],
                'best_val_loss': best_val_loss,
                'test_rmse': test_rmse,
                'train_rmse': train_rmse,
                'test_r2': test_r2,
                'train_r2': train_r2,
                'loss_curve': train_losses
            })
            key = f"h{hidden_dim}_lr{lr}_{act}"
            training_loss_curves[key] = train_losses

# Rank by best test RMSE (lowest first) and R^2 (highest first)
results_sorted_rmse = sorted(results, key=lambda d: d['test_rmse'])
results_sorted_r2 = sorted(results, key=lambda d: d['test_r2'], reverse=True)

print("\nTop 5 hyperparameter configs by lowest test RMSE:")
for i, res in enumerate(results_sorted_rmse[:5], 1):
    print(f"{i}. Hidden dim: {res['hidden_dim']}, lr: {res['lr']}, act: {res['activation']}, "
          f"Test RMSE: {res['test_rmse']:.4f}, Test R^2: {res['test_r2']:.4f}, Best val loss: {res['best_val_loss']:.5f}")

print("\nTop 5 hyperparameter configs by highest test R^2:")
for i, res in enumerate(results_sorted_r2[:5], 1):
    print(f"{i}. Hidden dim: {res['hidden_dim']}, lr: {res['lr']}, act: {res['activation']}, "
          f"Test RMSE: {res['test_rmse']:.4f}, Test R^2: {res['test_r2']:.4f}, Best val loss: {res['best_val_loss']:.5f}")

# Plot training loss for all hyperparameter combinations
plt.figure(figsize=(16, 10))
for res in results:
    key = f"h{res['hidden_dim']}_lr{res['lr']}_{res['activation']}"
    label = f"h={res['hidden_dim']}, lr={res['lr']:.0e}, act={res['activation']}"
    plt.plot(training_loss_curves[key], label=label, alpha=0.7)
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.title("Training Loss Curves for ALL Hyperparameter Combinations")
plt.legend(fontsize=9, loc='upper right', bbox_to_anchor=(1.35, 1.0))
plt.grid(True)
plt.tight_layout()
plt.show()

# Show config with lowest (best) val loss (may also appear in above lists)
best_result = min(results, key=lambda d: d['best_val_loss'])
print("\nBest hyperparameters by validation loss:")
print(f"  Hidden dim: {best_result['hidden_dim']}")
print(f"  Learning rate: {best_result['lr']}")
print(f"  Activation: {best_result['activation']}")
print(f"  Best val loss: {best_result['best_val_loss']:.5f}")
print(f"  Train RMSE: {best_result['train_rmse']:.4f}")
print(f"  Train R^2: {best_result['train_r2']:.4f}")
print(f"  Test RMSE:  {best_result['test_rmse']:.4f}")
print(f"  Test R^2:  {best_result['test_r2']:.4f}")
```
